```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense

# 1. Dataset Construction
sentences = [
    "I love this product, it is amazing!",
    "Great service and wonderful experience.",
    "Best purchase I have made in a long time.",
    "Absolutely terrible, I hate it.",
    "Worst quality ever, completely broken.",
    "Horrible experience, do not buy this.",
    "Fantastic quality, highly recommended!",
    "Very bad, waste of money.",
    "Exceeded my expectations, super happy!",
    "Disappointing and completely useless."
]
# 1 = Positive, 0 = Negative
labels = np.array([1, 1, 1, 0, 0, 0, 1, 0, 1, 0])

# 2. Text Tokenization and Vectorization
vocab_size = 100
embedding_dim = 16
max_length = 10
oov_tok = "<OOV>"

tokenizer = Tokenizer(num_words=vocab_size, oov_token=oov_tok)
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='post', truncating='post')

# 3. Model Architecture
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    GlobalAveragePooling1D(),  # Aggregates word vectors into a single sentence vector
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# 4. Training
epochs = 30
history = model.fit(padded_sequences, labels, epochs=epochs, verbose=1)

# 5. Visualization
plt.figure(figsize=(12, 4))

# Loss Plot
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Loss', color='red')
plt.title('Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.grid(True)

# Accuracy Plot
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Accuracy', color='blue')
plt.title('Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.grid(True)

plt.tight_layout()
plt.show()
```
