```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Step 1: Create a simple customer behavioral dataset
data = {
    'Age': [22, 25, 47, 52, 46, 56, 23, 40, 60, 32],
    'Income': [25000, 32000, 70000, 110000, 52000, 95000, 18000, 62000, 105000, 48000],
    'Previous_Purchases': [2, 3, 12, 25, 8, 18, 1, 9, 22, 5],
    'Buying_Range': ['Low', 'Low', 'Medium', 'High', 'Medium', 'High', 'Low', 'Medium', 'High', 'Low']
}

df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[['Age', 'Income', 'Previous_Purchases']]
y = df['Buying_Range']

# Step 2: Implement and Train the Decision Tree Classifier
clf = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=42)
clf.fit(X, y)

# Step 3: Display the Decision Tree Visualization
plt.figure(figsize=(12, 8))
plot_tree(
    clf,
    feature_names=X.columns,
    class_names=clf.classes_,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Customer Buying Range Decision Tree")
plt.show()
```
