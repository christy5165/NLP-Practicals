```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import bigrams

# Download required NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')

# Input sentence
text = "Machine learning is transforming industries."

# Step 1: Tokenize the text into individual words
tokens = word_tokenize(text)

# Step 2: Generate bigrams (pairs of consecutive words) using NLTK
bigrams_list = list(bigrams(tokens))

# Step 3: Display the generated bigrams
print("Sentence:", text)
print("\nGenerated Bigrams:")
print(bigrams_list)
```
