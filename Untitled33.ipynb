```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.util import ngrams

# Download required NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')

# Input sentence
text = "I love natural language processing."

# Step 1: Tokenize the text into individual words
tokens = word_tokenize(text)

# Step 2: Generate unigrams (n=1) using NLTK's ngrams function
unigrams_list = list(ngrams(tokens, 1))

# Step 3: Display the generated unigrams
print("Sentence:", text)
print("\nGenerated Unigrams:")
print(unigrams_list)
```
