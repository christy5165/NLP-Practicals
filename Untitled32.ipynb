```python
import spacy

# Load spaCy's English core model
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = "Apple Inc. was founded by Steve Jobs in Cupertino, California."

# Process text through spaCy pipeline
doc = nlp(text)

# 1. Display All Tokens with POS Tags
print("--- ALL TOKENS & POS TAGS ---")
print(f"{'Token':<15} | {'POS Tag (pos_)':<15} | {'Fine Tag (tag_)'}")
print("-" * 50)
for token in doc:
    print(f"{token.text:<15} | {token.pos_:<15} | {token.tag_}")

# 2. Extract Nouns (PROPN for Proper Nouns, NOUN for Common Nouns)
print("\n--- EXTRACTED NOUNS ---")
print(f"{'Noun':<15} | {'Type':<15} | {'POS Tag'}")
print("-" * 45)
for token in doc:
    if token.pos_ in ["NOUN", "PROPN"]:
        print(f"{token.text:<15} | {token.pos_:<15} | {token.tag_}")

# 3. Extract Named Entities (Persons, Locations, Organizations)
print("\n--- NAMED ENTITIES ---")
print(f"{'Entity Text':<20} | {'Label (Category)':<15} | {'Description'}")
print("-" * 60)
for ent in doc.ents:
    print(f"{ent.text:<20} | {ent.label_:<15} | {spacy.explain(ent.label_)}")
```
