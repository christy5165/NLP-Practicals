```python
import numpy as np
import pandas as pd
import gensim.downloader as api
from gensim.models import Word2Vec
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from nltk.tokenize import word_tokenize
import nltk

# Download tokenizer data
nltk.download('punkt')
nltk.download('punkt_tab')

# Step 1: Create a synthetic product reviews dataset
data = {
    'review': [
        "This product is amazing and high quality! Highly recommended.",
        "Worst purchase ever. Completely useless and broke immediately.",
        "Excellent battery life and very durable build.",
        "Terrible customer service and defective product.",
        "Works perfectly! Very satisfied with performance.",
        "Poor quality, stopped working within two days.",
        "Great value for money, functions as expected.",
        "Disappointing build quality and slow shipping."
    ],
    'sentiment': [1, 0, 1, 0, 1, 0, 1, 0]  # 1: Positive, 0: Negative
}

df = pd.DataFrame(data)

# Preprocessing: Tokenize reviews into lists of lowercased words
corpus = [word_tokenize(review.lower()) for review in df['review']]

# Step 2: Fine-Tune Word2Vec Embeddings
# Option A: Train a custom Word2Vec model on domain corpus (or update an existing gensim model)
vector_size = 100
w2v_model = Word2Vec(sentences=corpus, vector_size=vector_size, window=5, min_count=1, workers=4, epochs=30)

# Function to generate document vector by averaging word vectors
def get_document_vector(tokens, model, vector_size):
    vectors = [model.wv[word] for word in tokens if word in model.wv]
    if len(vectors) == 0:
        return np.zeros(vector_size)
    return np.mean(vectors, axis=0)

# Generate feature matrix X
X = np.array([get_document_vector(tokens, w2v_model, vector_size) for tokens in corpus])
y = np.array(df['sentiment'])

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Step 3: Train Sentiment Classifier using fine-tuned embeddings
classifier = LogisticRegression()
classifier.fit(X_train, y_train)

# Predict and Evaluate
y_pred = classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Test Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
