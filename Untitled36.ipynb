```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Create a customer behavioral dataset
data = {
    'Age': [22, 25, 47, 52, 46, 56, 23, 40, 60, 32, 28, 50, 38, 45, 29],
    'Income': [25000, 32000, 70000, 110000, 52000, 95000, 18000, 62000, 105000, 48000, 35000, 88000, 58000, 75000, 41000],
    'Time_Spent': [15, 25, 30, 10, 45, 5, 20, 35, 12, 18, 22, 14, 28, 32, 19],
    'Previous_Purchases': [2, 3, 12, 25, 8, 18, 1, 9, 22, 5, 3, 15, 7, 11, 4],
    'Buying_Range': ['Low', 'Low', 'Medium', 'High', 'Medium', 'High', 'Low', 'Medium', 'High', 'Low', 'Low', 'High', 'Medium', 'Medium', 'Low']
}

df = pd.DataFrame(data)

# Features (X) and Target (y)
X = df[['Age', 'Income', 'Time_Spent', 'Previous_Purchases']]
y = df['Buying_Range']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 2: Default Decision Tree Model (Baseline)
base_clf = DecisionTreeClassifier(random_state=42)
base_clf.fit(X_train, y_train)
y_pred_base = base_clf.predict(X_test)
base_acc = accuracy_score(y_test, y_pred_base)

# Step 3: Hyperparameter Tuning using GridSearchCV
param_grid = {
    'max_depth': [2, 3, 4, 5, None],
    'min_samples_split': [2, 3, 5],
    'criterion': ['gini', 'entropy']
}

grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=3,
    scoring='accuracy'
)
grid_search.fit(X_train, y_train)

# Best Model Evaluation
best_clf = grid_search.best_estimator_
y_pred_tuned = best_clf.predict(X_test)
tuned_acc = accuracy_score(y_test, y_pred_tuned)

# Display Evaluation Results
print("--- EVALUATION RESULTS ---")
print(f"Baseline Accuracy: {base_acc * 100:.2f}%")
print(f"Tuned Model Accuracy: {tuned_acc * 100:.2f}%")
print("\nOptimal Hyperparameters:", grid_search.best_params_)
print("\nClassification Report (Tuned Model):\n")
print(classification_report(y_test, y_pred_tuned))
```
