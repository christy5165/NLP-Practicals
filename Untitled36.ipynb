```python
import os
import spacy
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity

# 1. Download and load spaCy model with real word vectors
os.system("python -m spacy download en_core_web_md")
nlp = spacy.load("en_core_web_md")

# 2. Define words and extract vectors (Lines 9 & 13 fixed)
words = ["dog", "cat", "animal", "car", "bus"]
vectors = np.array([nlp(word).vector for word in words])

# 3. Compute pairwise cosine similarity matrix
similarity_matrix = cosine_similarity(vectors)

# 4. Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    similarity_matrix,
    annot=True,
    xticklabels=words,
    yticklabels=words,
    cmap="Blues",
    fmt=".3f"
)

plt.title("Word Embedding Cosine Similarity Matrix", fontsize=14)
plt.tight_layout()
plt.show()
```
