```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Step 1: Create a synthetic customer dataset with missing values and categorical data
data = {
    'Age': [25, 30, np.nan, 45, 35, 50, 23, np.nan, 40, 28],
    'Income': [50000, 60000, 75000, 80000, np.nan, 120000, 45000, 62000, 90000, 52000],
    'Time_Spent_Website': [15, 25, 30, 10, 45, 5, 20, 35, 12, 18],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'Buying_Range': ['Low', 'Low', 'Medium', 'High', 'Medium', 'High', 'Low', 'Medium', 'High', 'Low']
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)

# Step 1a: Handle Missing Values (Impute numerical columns with mean)
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Income'].fillna(df['Income'].mean(), inplace=True)

# Step 1b: Encode Categorical Data
le_gender = LabelEncoder()
df['Gender'] = le_gender.fit_transform(df['Gender'])

le_target = LabelEncoder()
df['Buying_Range_Encoded'] = le_target.fit_transform(df['Buying_Range'])

# Feature Matrix (X) and Target Vector (y)
X = df[['Age', 'Income', 'Time_Spent_Website', 'Gender']]
y = df['Buying_Range_Encoded']

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 2: Train a Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

# Step 3: Make predictions and evaluate performance using Accuracy
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("\nEvaluation Metrics:")
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
